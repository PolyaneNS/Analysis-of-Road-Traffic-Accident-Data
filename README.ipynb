# **📌 About the Project**
*This project performs an Exploratory Data Analysis (EDA) using accident records from the Brazilian Federal Highway Police (PRF), with the goal of identifying patterns that may contribute to accident prevention and support data-driven decision-making.*
# **🎯 Objectives**
*The analysis aims to answer the following questions:*

*   Which states recorded the highest number of accidents?
*   What are the main causes of accidents?
*   On which days of the week and at what times do accidents occur most frequently?
*   How do weather conditions influence accident occurrence?
*   Which types of accidents have the highest incidence and severity?

# **📊 Key Findings**


*   Minas Gerais, Santa Catarina, and Paraná recorded the highest number of accidents.
*   Most accidents occurred under clear weather conditions.
*   Accidents were concentrated mainly on weekends, with a peak between 5 PM and 7 PM.
*   The most frequent causes were delayed or ineffective driver reaction and incompatible speed.
*   Rear-end collisions and roadway departures were the most common accident types, while head-on collisions resulted in the highest number of fatalities.
