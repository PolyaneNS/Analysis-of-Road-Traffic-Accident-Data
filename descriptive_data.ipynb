```python
from google.colab import drive
import pandas as pd

# Monta o Google Drive
drive.mount('/content/drive')

import pandas as pd

df = pd.read_csv(
    "/content/drive/MyDrive/Curso_Data_Science/3_semana/datatran2026.csv",
    sep=";",
    encoding="latin1",
    decimal=","
)
```

```python

# Understanding the dataset

print(f"Linhas | Rows: {df.shape[0]:,}")
print(f"Colunas | Columns: {df.shape[1]}")
print(f"Duplicados | Duplicates: {df.duplicated().sum():,}")

resumo_dados = pd.DataFrame({
    "Tipo": df.dtypes,
    "Não nulos": df.notna().sum(),
    "Nulos": df.isna().sum(),
    "Valores únicos": df.nunique()
})

display(resumo_dados)

```

```python

# Descriptive statistics for numerical variables
display(df.describe().T.round(2))
```

```python

# Dataset time range
# Converts the date column to datetime format
df["data_inversa"] = pd.to_datetime(
    df["data_inversa"],
    dayfirst=True,
    errors="coerce"
)

# Identifies the time period covered by the dataset
data_inicial = df["data_inversa"].min()
data_final = df["data_inversa"].max()

duracao_dias = (data_final - data_inicial).days

resumo_periodo = pd.DataFrame({
    "Informação | Information": [
        "Data inicial | Start date",
        "Data final | End date",
        "Período total | Total period"
    ],
    "Valor | Value": [
        data_inicial.strftime("%d/%m/%Y"),
        data_final.strftime("%d/%m/%Y"),
        f"{duracao_dias:,} dias"
    ]
})

display(resumo_periodo)
```
